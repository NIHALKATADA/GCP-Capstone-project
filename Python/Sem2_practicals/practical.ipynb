```python
def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    data = sorted(data)
    n = len(data)
    if n % 2 == 1:
        return data[n // 2]
    else:
        return (data[n // 2 - 1] + data[n // 2]) / 2

def calculate_mode(data):
    freq = {}
    for value in data:
        freq[value] = freq.get(value, 0) + 1
    max_freq = max(freq.values())
    modes = [key for key, value in freq.items() if value == max_freq]
    return modes

def calculate_variance(data):
    # Safety check: Variance requires at least 2 numbers for Sample Variance
    if len(data) < 2:
        return 0.0
    mean = calculate_mean(data)
    return sum((x - mean) ** 2 for x in data) / (len(data) - 1)

def calculate_std_dev(data):
    return calculate_variance(data) ** 0.5

def calculate_range(data):
    return max(data) - min(data)

# -------- Main Program --------
try:
    user_input = input("Enter numbers separated by space: ")
    # Check if input is empty
    if not user_input.strip():
        print("Error: No numbers entered.")
    else:
        data = list(map(int, user_input.split()))
        
        print("-" * 30)
        print(f"Data: {data}")
        print("Mean:", calculate_mean(data))
        print("Median:", calculate_median(data))
        print("Mode:", calculate_mode(data))
        print("Variance:", calculate_variance(data))
        print("Standard Deviation:", calculate_std_dev(data))
        print("Range:", calculate_range(data))
        print("-" * 30)

except ValueError:
    print("Error: Please enter only valid integers separated by spaces.")
```
