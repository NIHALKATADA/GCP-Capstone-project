1. Write a Python program to read a list of student names and display only those names whose length is greater than 5.

```python
student_names = []
print("Enter student names (type 'done' to finish):")
while True:
    entry = input("> ")
    if entry.lower() == 'done':
        break
    student_names.append(entry)
print("\nNames with length greater than 5:")
print("-" * 30)
# Using a list comprehension for a concise solution
long_names = [name for name in student_names if len(name) > 5]
for name in long_names:
    print(name)
```

2.Write a Python program
 A. take input of string
 B. reverse a given string without using built-in function
 C . Check if it is a palindrome

```python
text = input("Enter a string: ")
reversed_text = ""
length = len(text)
for i in range(length - 1, -1, -1):
    reversed_text += text[i]
print("Reversed String:", reversed_text)
if text.lower() == reversed_text.lower():
    print("It is a Palindrome")
else:
    print("It is NOT a Palindrome")
```

```python
text = input("Enter a string: ")
reversed_text =text[::-1]
print("Reversed String:", reversed_text)
if text.lower() == reversed_text.lower():
    print("It is a Palindrome")
else:
    print("It is NOT a Palindrome")
```

3. Write a Python program to create a dictionary of student names and marks and display students who scored above 85.

```python
student_marks = {}
print("Enter student details (type 'done' as name to stop):")
while True:
    name = input("Enter student name: ")
    if name.lower() == 'done':
        break
    marks = int(input(f"Enter marks for {name}: "))
    student_marks[name] = marks
print("\nStudents who scored above 85:")
print("-" * 30)
found = False
for name, marks in student_marks.items():
    if marks > 85:
        print(f"{name}: {marks}")
        found = True
if not found:
    print("No students scored above 85.")
```

4. Take input of numbers and check how many even and odd numbers.

```python
list=[]
choice=int(input("How many number you want to enter?"))
for i in range(choice):
    num=int(input("Enter The Numbers"))
    list.append(num)
CountEven=0
CountOdd=0
for i in list:
    if i%2==0:
        CountEven+=1
    else:
        CountOdd+=1
print(f"Even Numbers= {CountEven}\n Odd Numbers= {CountOdd}")

```
