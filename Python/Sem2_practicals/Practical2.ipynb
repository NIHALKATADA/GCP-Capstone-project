1. Perform matrix multiplication of two 2×2 matrices using numpy.

```python
import numpy as np
matrix_a = np.array([[1, 2], 
                     [3, 4]])
matrix_b = np.array([[5, 6], 
                     [7, 8]])
result = np.dot(matrix_a, matrix_b)
print("Matrix A:\n", matrix_a)
print("\nMatrix B:\n", matrix_b)
print("\nProduct (A x B):\n", result)
```

2. Add a new column to DataFrame using pandas

```python
import pandas as pd

# 1. Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [20, 21, 19]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
df['Grade'] = ['A', 'B', 'A']

print("\nDataFrame with New Column:")
print(df)
```

3. Fetch Data from Public API (Requests)

```python
import requests
url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url)
if response.status_code == 200:
    data = response.json()  # Parse JSON response
    print("Data Fetched Successfully:")
    print(f"Title: {data['title']}")
    print(f"Body: {data['body']}")
else:
    print(f"Failed to retrieve data. Status code: {response.status_code}")
```

4. Case Study: Student Performance Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import io
csv_content = """Name,Marks,Attendance
Rohan,85,90
Priya,92,95
Amit,NaN,80
Sneha,78,85
Vikram,65,70
Anjali,88,NaN
"""
with open('student_data.csv', 'w') as f:
    f.write(csv_content)
print("--- Case Study Output ---")
df = pd.read_csv('student_data.csv')
print("\n1. Original Dataset:\n", df)
df_clean = df.dropna()
print("\n2. Dataset after removing missing values:\n", df_clean)
average_marks = df_clean['Marks'].mean()
print(f"\n3. Average Marks: {average_marks:.2f}")
above_avg_students = df_clean[df_clean['Marks'] > average_marks]
print("\n4. Students with Above Average Marks:\n", above_avg_students)
plt.figure(figsize=(8, 5))
plt.bar(df_clean['Name'], df_clean['Marks'], color='skyblue')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.title('Student Marks Distribution')
plt.axhline(y=average_marks, color='r', linestyle='--', label=f'Average ({average_marks:.1f})')
plt.legend()
plt.show()
```
