SAT Solver Using Brute force

```python
import itertools

def eval_clause(clause, assignment):
    for literal in clause:
        var = abs(literal)
        val = assignment[var]
        
        if (literal > 0 and val) or (literal < 0 and not val):
            return True
    return False

def sat_solver(cnf, n_vars):
    for values in itertools.product([False, True], repeat=n_vars):
        assignment = {i+1: values[i] for i in range(n_vars)}
        
        if all(eval_clause(clause, assignment) for clause in cnf):
            return "SATISFIABLE", assignment
            
    return "UNSATISFIABLE", None

if __name__ == "__main__":
    cnf = [
        [1, -2],
        [2, 3]
    ]

    result, assignment = sat_solver(cnf, 3)

    print(result)
    if assignment:
        print("Assignment:", assignment)
```
