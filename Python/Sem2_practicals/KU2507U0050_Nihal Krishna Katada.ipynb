NQueen (Part 2)

```python
def solve_n_queens(n):
    solutions = []
    # Create an empty N x N board filled with dots
    board = [["." for _ in range(n)] for _ in range(n)]
    # Helper function to check if placing a queen here is safe
    def is_safe(row, col):
        # 1. Check the row to the left
        for i in range(col):
            if board[row][i] == "Q":
                return False
                
        # 2. Check the upper diagonal to the left
        for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
            if board[i][j] == "Q":
                return False
                
        # 3. Check the lower diagonal to the left
        for i, j in zip(range(row, n, 1), range(col, -1, -1)):
            if board[i][j] == "Q":
                return False
                
        return True

    # Core backtracking function
    def backtrack(col):
        # Base Case: If we successfully placed queens in all columns, save the board
        if col == n:
            valid_board = [" ".join(r) for r in board]
            solutions.append(valid_board)
            return

        # Try placing a queen in each row of the current column
        for row in range(n):
            if is_safe(row, col):
                board[row][col] = "Q"    # Place the queen
                backtrack(col + 1)       # Move to the next column
                board[row][col] = "."    # BACKTRACK: Remove the queen and try the next row

    # Start the process at the very first column (index 0)
    backtrack(0)
    return solutions

# Let's run it for a 4x4 board
n = 4
results = solve_n_queens(n)

print(f"Found {len(results)} solutions for {n}-Queens:\n")
for idx, solution in enumerate(results):
    print(f"Solution {idx + 1}:")
    for row in solution:
        print(row)
    print("-" * 10)
```
