Case Study 1 :
Finding Friends on a Social Network using BFS
Each person is a node.
Connections represent friendships.

```python
from collections import deque
def bfs_social_network(graph, start_node):
    visited = set()
    queue = deque([start_node])
    visited.add(start_node)
    print(f"Starting BFS from: {start_node}")
    while queue:
        current_person = queue.popleft()
        print(f"Visiting: {current_person}")
        for friend in graph[current_person]:
            if friend not in visited:
                visited.add(friend)
                queue.append(friend)
social_network = {
    'Alice': ['Bob', 'Charlie'],
    'Bob': ['Alice', 'David', 'Eve'],
    'Charlie': ['Alice', 'Frank'],
    'David': ['Bob'],
    'Eve': ['Bob', 'Frank'],
    'Frank': ['Charlie', 'Eve']
}
bfs_social_network(social_network, 'Alice')
```

Case Study 2 :
Website Page Traversal (DFS)
A crawler visits web pages deeply.

```python
def dfs_web_crawler(graph, current_page, visited=None):
    if visited is None:
        visited = set()
    visited.add(current_page)
    print(f"Crawling: {current_page}")
    for link in graph[current_page]:
        if link not in visited:
            dfs_web_crawler(graph, link, visited)
web_structure = {
    'Home': ['About', 'Contact'],
    'About': ['Team', 'History'],
    'Team': ['CEO', 'CTO'],
    'History': [],
    'Contact': ['Map'],
    'Map': [],
    'CEO': [],
    'CTO': []
}
print("Starting Web Crawler (DFS):")
dfs_web_crawler(web_structure, 'Home')
```
