```python
import pandas as pd
import numpy as np
```

```python
data = [12, 15, 14, 10, 18, 20, 15, 22, 25, 15, 30, 10, 14, 19, 21]
df = pd.DataFrame(data, columns=['Values'])
mean_val = df['Values'].mean()
median_val = df['Values'].median()
mode_val = df['Values'].mode()[0]
std_dev = df['Values'].std()
variance = df['Values'].var()
min_val = df['Values'].min()
max_val = df['Values'].max()
data_range = max_val - min_val
q1 = df['Values'].quantile(0.25) # 25th percentile
q3 = df['Values'].quantile(0.75) # 75th percentile
iqr = q3 - q1
```

```python

print("--- Descriptive Statistical Measures Report ---")
print(f"Dataset Size: {len(df)}")
print(f"   Mean (Average): {mean_val:.2f}")
print("\n--- Summary using Describe Function ---")
print(df.describe())
print(f"var     {variance}")
print(f"Datarange {data_range}")
print(f"iqr     {iqr}")
```
